## 19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

**Explicação:** Calculei o produto escalar, normalizei pelos módulos dos vetores e depois usei arccos para obter o ângulo.

```python
# --- Exercício 19 ---
import numpy as np
a = np.random.randint(1, 100, size=20)
b = np.random.randint(1, 100, size=20)

cos_theta = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
cos_theta = np.clip(cos_theta, -1.0, 1.0)

angulo_rad = np.arccos(cos_theta)
angulo_deg = np.degrees(angulo_rad)

print(f"Ângulo em radianos: {angulo_rad:.4f}")
print(f"Ângulo em graus: {angulo_deg:.2f}°")
```

## 7. Crie uma nova coluna chamada Total que seja o produto entre as colunas Preço e Quantidade.

## 11. Preencha os valores nulos de uma coluna numérica com a mediana dessa coluna.
